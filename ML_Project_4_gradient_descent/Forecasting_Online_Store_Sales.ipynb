```python
import numpy as np
import pandas as pd
df_ads=pd.read_csv('advertising.csv')
df_ads.head() #this will display the first 5 rows of the dataframe

import matplotlib.pyplot as plt
import seaborn as sns #Seaborn is a Python data visualization library based on matplotlib. 
sns.heatmap(df_ads.corr(),cmap='coolwarm',annot=True)
plt.show()
# Now, we can see the correlation between the features and the target variable.
# In our case, we can see wechat and sales have a strong positive correlation, 0.9.
# This means that as wechat increases, sales also increases.

# height is the height of each plot, aspect is the aspect ratio of each plot, kind is the type of plot, in this case scatter plot.
sns.pairplot(df_ads,x_vars=['wechat','weibo','others'],y_vars='sales',height=5,aspect=1,kind='scatter')

plt.show()
```

```python
x=np.array(df_ads.wechat)
y=np.array(df_ads.sales)
print('Dimension of x:', x.ndim)
print('shape of x:', x.shape) #(200,) means 200 rows and 1 column, this is 1D array
print('values of x:', x)
```

```python
# For numerical datasets with regression problems, the standard format is to have the features in a 2D array with shape (samples, features). 
# In this case, we have 200 samples and 1 feature, so we need to reshape x to be a 2D array with shape (200, 1).
x=x.reshape((len(x),1))
y=y.reshape((len(y),1))
print('Dimension of x after reshape:', x.ndim)
print('shape of x after reshape:', x.shape) #(200,1) means 200 rows and 1 column, this is 2D array
print('values of x after reshape:', x)
```

```python
# Split the data into training and testing sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)

# Normalization is a technique to scale the data to a specific range, usually [0, 1].
def scaler(train,test):
    min = train.min(axis=0) 
    max = train.max(axis=0)
    gap = max - min # Calculate the range of the training set
    train -= min #All the values in the training set are subtracted by the minimum value of the training set.
    train /= gap #All the values in the training set are divided by the range of the training set.
    test -= min #All the values in the testing set are subtracted by the minimum value of the training set.
    test /= gap #All the values in the testing set are divided by the range of the training set.
    return train,test

x_train,x_test=scaler(x_train,x_test) #Normalize the training and testing sets using the scaler function defined above.
y_train,y_test=scaler(y_train,y_test) #Normalize the training and testing sets using the scaler function defined above.

# Now, we can visualize the training set and testing set using a scatter plot.
plt.plot(x_train,y_train,'r.',label='Training set') #Plot the training set in red color with label 'Training set'
plt.xlabel('WeChat') #Set the x-axis label to 'WeChat'
plt.ylabel('Sales') #Set the y-axis label to 'Sales'
plt.legend() #Display the legend
plt.show() #Display the plot

```

```python
def loss_function(x,y,weight,bias):
    y_hat = weight*x+bias
    loss = y_hat-y
    cost = np.sum(loss**2)/(2*len(x))
    return cost

print("when the weight is 5, and bias is 3, the loss will be:", loss_function(x_train,y_train,weight=5,bias=3))
print("when the weight is 100, and bias is 1, the loss will be:",  loss_function(x_train,y_train,weight=100,bias=1))

# Now, we can see y=3x+5 is better than y=100x+1.
# But there is not enough, we need to use gradient descent.
def gradient_descent(x,y,w,b,alpha,iter): #alpha is the learning rate.
    l_history = np.zeros(iter) #Initialize an array to track the loss during the gradient descent process
    w_history = np.zeros(iter) #Initialize an array to track the weights during the gradient descent process
    b_history = np.zeros(iter) #Initialize an array to track the bias during the gradient descent process
    for i in range(iter):
        y_hat = w*x+b
        loss = y_hat-y
        derivative_w = x.T.dot(loss)/len(x)
        derivative_b = sum(loss)*1/len(x)
        w = w - alpha*derivative_w
        b = b - alpha*derivative_b
        l_history[i] = loss_function(x,y,w,b)
        w_history[i] = w
        b_history[i] = b

    return l_history,w_history,b_history

#Initial 
iterations = 100
alpha = 1
weight = -5
bias = 3
print("loss:",loss_function(x_train,y_train,weight,bias))

plt.plot(x_train,y_train,'r.',label = 'Training data') #show the training data.
line_x = np.linspace(x_train.min(),x_train.max(),500)
line_y = [weight*x+bias for x in line_x]
plt.plot(line_x,line_y,'b--',label = 'current hypothesis')
plt.xlabel('wechat')
plt.ylabel('sales')
plt.legend()
plt.show()


```

```python
#start gradient descent
loss_history,weight_history,bias_history = gradient_descent(x_train,y_train,weight,bias,alpha,iterations)
plt.plot(loss_history,'g--',label='Loss curve')# display loss curve
plt.xlabel('Iterations')
plt.ylabel('Loss')
plt.legend()
plt.show()

#draw the current model
plt.plot(x_train,y_train,'r.',label='Training data')
line_x = np.linspace(x_train.min(),x_train.max(),500)
line_y = [weight_history[-1]*xx+bias_history[-1] for xx in line_x]
plt.plot(line_x,line_y,'b--',label = 'current hypothesis')
plt.xlabel('wechat')
plt.ylabel('sales')
plt.legend()
plt.show()


print('current loss:', loss_function(x_train,y_train,weight_history[-1],bias_history[-1]))
print('current weight:',weight_history[-1])
print('current bias:',bias_history[-1])

print('loss in test sets:',loss_function(x_test,y_test,weight_history[-1],bias_history[-1]))

#Compare the training sets and testing sets loss curves.
test_loss_history,test_weight_history,test_bias_history = gradient_descent(x_test,y_test,weight,bias,alpha,iterations)
loss_history,weight_history,bias_history = gradient_descent(x_train,y_train,weight,bias,alpha,iterations)
plt.plot(loss_history,'r--',label='Training sets loss curve')# display train loss curve
plt.plot(test_loss_history,'b--',label='Testing sets loss curve')
plt.xlabel('Iterations')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
